```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("games.csv")
df
```

```python
df = df.drop(columns=["Metacritics"])
df
```

```python
df["Popularity"] = pd.cut(
    df["Steam_Total_Reviews"],
    bins=[0, 500, 2000, 10000, float("inf")],
    labels=["Niche", "Moderate", "Popular", "Blockbuster"],
    right=False,
).astype(str)

#df = df.drop(columns=["Steam_Total_Reviews"])
print(df["Popularity"].value_counts())
```

```python
df.head()
```

```python
df["Game_Genre"] = df["Game_Genres_All"].str.split("|").str[0]
df.loc[df["Game_Genre"] == "", "Game_Genre"] = "Other"

print(df["Game_Genre"].value_counts())
```

```python
df["Game_Genre"] = df["Game_Genre"].replace({
    "Racing": "Sports & Racing",
    "Sports": "Sports & Racing",
    "Casual": "Other",
    "Massively Multiplayer": "Other",
})
print(df["Game_Genre"].value_counts())
```

```python
df.head()
```

```python
df = df.drop(columns=["Game_Genres_All","Steam_Total_Reviews"])
```

```python
df.to_csv("cleaned games.csv", index=False)
```
