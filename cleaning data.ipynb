```python
import pandas as pd
import matplotlib
```

```python
df_G = pd.read_csv("excel files\games.csv")
df_G.head()
```

```python
df_R = pd.read_csv("excel files\steam_reviews.csv")
df_R.head()
```

```python
df_P = pd.read_csv("excel files\steam_prices.csv")
df_P.head()
```

```python
df_P = df_P.drop(columns="Currency")
df_P.head()
```

```python
df = pd.merge(df_G, df_R, on="GameID", how="inner")
df.head()
```

```python
df = pd.merge(df, df_P, on="GameID", how="inner")
df.head()
```

```python
df = df.drop(columns=["GameID","MetricID","PriceID","Steam_AppID_x","Steam_AppID_y","Fetched_At","Updated_At"])
df
```

```python
print("Duplicate games count:", df.duplicated(subset=["Game_Name"]).sum())
print(df[df.duplicated(subset=["Game_Name"], keep=False)])
```

```python
df = df.dropna(subset=['Steam_Rating_Pct']).reset_index(drop=True)
```
