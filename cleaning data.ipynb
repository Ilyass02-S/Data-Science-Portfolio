```python
import pandas as pd
import matplotlib
```

```python
df_G = pd.read_csv("excel files\games.csv")
df_G.head()
```

```python
df_R = pd.read_csv("excel files\steam_reviews.csv")
df_R.head()
```

```python
df_P = pd.read_csv("excel files\steam_prices.csv")
df_P.head()
```

```python
df = pd.merge(df_G, df_R, on="GameID", how="inner")
df.head()
```

```python
df = pd.merge(df, df_P, on="GameID", how="inner")
df.head()
```

```python
df = df.drop(columns=["GameID","MetricID","PriceID","Steam_AppID_x","Steam_AppID_y","Fetched_At","Updated_At"])
df
```

```python
df = df.drop_duplicates(subset=["Game_Name"], keep="first")
print("Duplicate games count remaining:", df.duplicated(subset=["Game_Name"]).sum())
```

```python
df = df.dropna(subset=['Steam_Rating_Pct']).reset_index(drop=True)
```

```python
df[df["Initial_Price"].isnull()]
```

```python
df["Metacritics"].isnull().sum()
```

```python
import numpy as np
```

```python
df["Metacritics"] = df["Metacritics"].fillna(df["Steam_Rating_Pct"])
```

```python
bins = [-1, 19, 39, 69, 79, 94, 100]
labels = [
    "Overwhelmingly Negative",
    "Mostly Negative",
    "Mixed",
    "Mostly Positive",
    "Very Positive",
    "Overwhelmingly Positive",
]

df["steam_rating_text"] = pd.cut(
    df["Steam_Rating_Pct"], bins=bins, labels=labels
)
df = df.drop(columns=["Steam_Rating_Pct"])
```
