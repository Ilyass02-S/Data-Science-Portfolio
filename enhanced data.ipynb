```python
import pandas as pd
import matplotlib
```

```python
df_G = pd.read_csv("excel files\games.csv")
df_G.head()
```

```python
df_R = pd.read_csv("excel files\steam_reviews.csv")
df_R.head()
```

```python
df_P = pd.read_csv("excel files\steam_prices.csv")
df_P.head()
```

```python
df = pd.merge(df_G, df_R, on="GameID", how="inner")
df.head()
```

```python
df = pd.merge(df, df_P, on="GameID", how="inner")
df.head()
```

```python
df = df.drop(columns=["GameID","MetricID","PriceID","Steam_AppID_x","Steam_AppID_y","Fetched_At","Updated_At"])
df
```

```python
df = df.drop_duplicates(subset=["Game_Name"], keep="first")
print("Duplicate games count remaining:", df.duplicated(subset=["Game_Name"]).sum())
```

```python
df = df.dropna(subset=['Steam_Rating_Pct']).reset_index(drop=True)
```

```python
df[df["Initial_Price"].isnull()]
```

```python
df["Metacritics"].isnull().sum()
```

```python
import numpy as np
```

```python
df["Metacritics"] = df["Metacritics"].fillna(df["Steam_Rating_Pct"])
```

```python
bins = [-1, 19, 39, 69, 79, 94, 100]
labels = [
    "Overwhelmingly Negative",
    "Mostly Negative",
    "Mixed",
    "Mostly Positive",
    "Very Positive",
    "Overwhelmingly Positive",
]

df["steam_rating_text"] = pd.cut(
    df["Steam_Rating_Pct"], bins=bins, labels=labels
)
df = df.drop(columns=["Steam_Rating_Pct"])
```

```python
df["RAWG_Rating"].isnull().sum()
```

```python
df[df["RAWG_Rating"].isnull()]
```

```python
# Impute using the median RAWG rating of the dataset
rawg_median = df["RAWG_Rating"].median()
df["RAWG_Rating"] = df["RAWG_Rating"].fillna(rawg_median)
df = df.drop(columns=["Added_Count","Ratings_Count"])
df
```

```python
log_reviews = np.log1p(df["Steam_Total_Reviews"])

df["Popularity_Index"] = (log_reviews - log_reviews.min()) / (
    log_reviews.max() - log_reviews.min()
)

# 4. Classify into explicit popularity tiers
bins = [-0.01, 0.35, 0.65, 0.85, 1.00]
labels = ["Niche", "Moderate", "Popular", "Blockbuster"]

df["Popularity"] = pd.cut(
    df["Popularity_Index"], bins=bins, labels=labels
)
df = df.drop(columns=["Steam_Total_Reviews","Popularity_Index"])
df
```

```python
def categorize_pricing(row):
    if pd.notnull(row["Discount_Pct"]) and row["Discount_Pct"] != 0:
        return "Offre"
    elif row["Is_Free"] == 0:
        return "Free"
    else:
        return "Paid"

df["Pricing"] = df.apply(categorize_pricing, axis=1)
df = df.drop(columns=["Is_Free"])
df
```

```python
df.to_csv("excel files\cleaned data.csv", index=False)
```
